### C. Langkah Latihan 1, 2 dan 3

```python
import pandas as pd
data = {'luas':[36,70,45,90,54], 'kamar':[2,3,2,4,3],
        'lokasi':['Pinggir','Tengah','Pinggir','Tengah','Pinggir'],
        'harga':[250,600,320,800,400]}  # LABEL
df = pd.DataFrame(data)
X = df[['luas','kamar','lokasi']]   # fitur
y = df['harga']                     # label

# Label numerik kontinu -> REGRESI
print(df['harga'].tolist())
# Label kategori -> KLASIFIKASI
df['kategori'] = ['Murah','Mahal','Murah','Mahal','Murah']

from sklearn.model_selection import train_test_split
X_train,X_test,y_train,y_test = train_test_split(
    X, y, test_size=0.4, random_state=42)
print('Latih:', len(X_train), ' Uji:', len(X_test))

```

### Refleksi

#### 1. Apa perbedaan supervised, unsupervised, dan reinforcement learning?

Jawab:
- Supervised Learning adalah metode machine learning yang model nya dilatih untuk belajar menggunakan data yang sudah berlabel atau sudah punya jawaban untuk mengetahui prediksi.
- Unsupervised Learning adalah metode machine yang model nya dilatih untuk belajar menggunakan data yang tidak berlabel untuk mengetahui struktur tersembunyi/mengetahui hubungan di dalam suatu data
- Reinforcement Learning adalah metode machine learning yang model nya dilatih untuk belajar melalui interaksi via reward dan punishment untuk memaksimalkan reward.

#### 2. Mengapa pembagian data latih dan uji penting?

Jawab:

Data latih digunakan untuk melatih model untuk mengetahui pola data nya, sedangkan data uji digunakan untuk menguji model pada data baru yang belum terlihat. Jadi, data latih dan data uji penting agar kita bisa mengetahui seberapa akurat model yang sudah dibuat terhadap data yang belum pernah dilihat.


#### 3. Mengapa alur kerja ML disebut iteratif?

Jawab:

Karena dalam alur kerja machine learning, model dilatih berkali-kali dengan tujuan untuk mencari akurasi terbaik. Bahkan, setelah tahap evaluasi pun biasanya sering kali kembali ke tahap sebelumnya jika akurasi modelnya rendah. Itulah mengapa alur kerja ML disebut iteratif atau berulang.
